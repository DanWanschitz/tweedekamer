```python
import requests
import pandas as pd
r_besluit = requests.get(
    "https://gegevensmagazijn.tweedekamer.nl/OData/v4/2.0/Besluit"
    "?$filter=Verwijderd eq false and "
    "(BesluitSoort eq 'Stemmen - aangenomen' or BesluitSoort eq 'Stemmen - verworpen')"
    "&$select=Id,Agendapunt_Id,BesluitSoort,BesluitTekst,StemmingsSoort,GewijzigdOp")
r_besluit.status_code
r_besluit.json()

data_besluit = r_besluit.json()["value"]

df_besluit = pd.DataFrame(data_besluit)

print(df_besluit.head())

df_besluit.to_csv("besluiten.csv", index=False)
```

```python
r_agendapunt = requests.get(
    "https://gegevensmagazijn.tweedekamer.nl/OData/v4/2.0/Agendapunt"
    "?$filter=Verwijderd eq false"
    "&$select=Id, Activiteit_Id, Onderwerp, Aanvangstijd, Eindtijd, Volgorde, Status"
)
r_agendapunt.status_code
r_agendapunt.json()

data_agendapunt = r_agendapunt.json()["value"]

df_agendapunt = pd.DataFrame(data_agendapunt)

print(df_agendapunt.head())

df_agendapunt.to_csv("agendapunten.csv", index=False)
```

```python
r_activiteit = requests.get(
    "https://gegevensmagazijn.tweedekamer.nl/OData/v4/2.0/Activiteit"
    "?$filter=Verwijderd eq false"
    "&$select=Id, Soort, Onderwerp, Datum, Aanvangstijd, Eindtijd, Status, Vergaderjaar, Kamer"
)

r_activiteit.status_code
r_activiteit.json()

data_activiteit = r_activiteit.json()["value"]

df_activiteit = pd.DataFrame(data_activiteit)

print(df_activiteit.head())

df_activiteit.to_csv("activiteiten.csv", index=False)
```
